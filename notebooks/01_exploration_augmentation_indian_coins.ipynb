```python
# Chargement des données
import os
from os import *
def get_image_paths(parent_directory, image_extensions=['.jpg', '.jpeg', '.png', '.gif']):
    image_paths = []

    for root, dirs, files in os.walk(parent_directory):
        for file in files:
            if any(file.lower().endswith(ext) for ext in image_extensions):
                image_paths.append(os.path.join(root, file))

    return image_paths

# Replace 'your_parent_directory' with the actual path of your parent directory

```

## 2. Augmentation des données

```python
import albumentations as A
from PIL import Image

# Define augmentation pipeline
transform = A.Compose([
    A.RandomRotate90(),  # Randomly rotate 90, 180, or 270 degrees
    A.HorizontalFlip(p=0.5),  # Randomly flip horizontally
    A.ColorJitter(brightness=0.1, contrast=0.2, saturation=0.2, hue=0.05),  # Adjust brightness, contrast, saturation, and hue
    ])
```

```python
import numpy as np
def data_augement(A,B, C, save_path, transform):
    l=C[A:B]
    for i in range(len(l)):
        input_path=l[i]
        image=Image.open(input_path)
        # Apply augmentations
        image_array = np.array(image)
        augmented_image = transform(image=image_array)['image']
        # Save augmented image
        image_name = os.path.basename(input_path)
        output_path=save_path+image_name[:-4]+'_augmented'+image_name[-4:]
        print( output_path)
        Image.fromarray(augmented_image).save(output_path)
```

```python
import albumentations as A
from PIL import Image

# Define augmentation pipeline
transform = A.Compose([
    A.RandomRotate90(),  # Randomly rotate 90, 180, or 270 degrees
    A.HorizontalFlip(p=1),  # Randomly flip horizontally
    A.ColorJitter(brightness=0.2, contrast=0.1, saturation=0.2, hue=0.05),  # Adjust brightness, contrast, saturation, and hue
    ])
```

```python
import numpy as np
def data_augement(A,B, C, save_path, transform):
    l=C[A:B]
    for i in range(len(l)):
        input_path=l[i]
        image=Image.open(input_path)
        # Apply augmentations
        image_array = np.array(image)
        augmented_image = transform(image=image_array)['image']
        # Save augmented image
        image_name = os.path.basename(input_path)
        output_path=save_path+image_name[:-4]+'_augmented_2'+image_name[-4:]
        print( output_path)
        Image.fromarray(augmented_image).save(output_path)
```
